### ** Handling Missing Values **  

```python
# ● Check for missing values
round(df.isnull().mean()*100,1)
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df.isnull().sum()
```

```python
#  Impute missing values

df['payment']=df['payment'].fillna(df['payment'].mode()[0])
```

```python
df['pickup_zone']=df['pickup_zone'].fillna(df['pickup_zone'].mode()[0])
```

```python
df['dropoff_zone']=df['dropoff_zone'].fillna(df['dropoff_zone'].mode()[0])
```

```python
df['pickup_borough']=df['pickup_borough'].fillna(df['pickup_borough'].mode()[0])
```

```python
df['dropoff_borough']=df['dropoff_borough'].fillna(df['dropoff_borough'].mode()[0])
```

### ** Visualizations using Matplotlib/Pandas Plot: **

```python
# Line Chart to visualize the fare over time, using the pickup timestamp:

df['pickup'] = pd.to_datetime(df['pickup'])
df = df.sort_values('pickup')
df.plot(x='pickup', y='fare', kind='line', figsize=(12,4))
plt.title('Fare Over Time')
plt.xlabel('Pickup Timestamp')
plt.ylabel('Fare ($)')
plt.show()
```

```python
# A bar chart to show the total fare for each pickup_borough.

pu_borough_fares = df.groupby('pickup_borough')['fare'].sum()
pu_borough_fares.plot(kind='bar', figsize=(12, 4), color='skyblue')
plt.title('Total Fare by Pickup Borough', fontsize=14, fontweight='bold')
plt.xlabel('Pickup Borough', fontsize=12)
plt.ylabel('Total Fare ($)', fontsize=12)
plt.xticks(rotation=360)
plt.show()
```

```python
# A pie chart showing the distribution of trips based on the payment method 

payment_counts = df['payment'].value_counts()
payment_counts.plot(kind='pie',figsize=(5,5),
                        autopct='%1.1f%%',  # Displays percentages automatically on the slices
                        startangle=140,     # Rotates the start of the pie chart for a cleaner look
                        colors=['#4CAF50', '#FFC107', '#00BCD4', '#E91E63'] # Clean, modern color palette
)
plt.title('Trip Distribution by Payment Method', fontsize=14, fontweight='bold')
plt.ylabel('')  # Removes the default 'payment' label from the y-axis for clean design
plt.tight_layout()
plt.show()
```

```python
# A histogram to visualize the distribution of distance.

df['distance'].plot(kind='hist', bins=50, range=(0, 30), figsize=(12,4), color='teal', edgecolor='black', alpha=0.8)
plt.title('Distribution of Trip Distances', fontsize=14, fontweight='bold')
plt.xlabel('Distance (miles)', fontsize=12)
plt.ylabel('Number of Trips (Frequency)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# A box plot to visualize the distribution of tip amounts for each pickup_borough.

borough_groups = [group['tip'].dropna() for name, group in df.groupby('pickup_borough')]
borough_names = [name for name, group in df.groupby('pickup_borough')]
plt.figure(figsize=(12, 4))
plt.boxplot(
    borough_groups, 
    label=borough_names, 
    showfliers=False,  # Hides extreme outliers so you can actually read the boxes
    patch_artist=True,  # Allows filling the boxes with color
    boxprops=dict(facecolor='lightblue', color='navy'),
    medianprops=dict(color='darkred', linewidth=2)
)
plt.title('Distribution of Tip Amounts by Pickup Borough', fontsize=14, fontweight='bold')
plt.xlabel('Pickup Borough', fontsize=12)
plt.ylabel('Tip Amount ($)', fontsize=12)
plt.xticks(rotation=360)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### ** Visualizations using Seaborn: **

```python
# A count plot to visualize the number of trips in each pickup_borough.

plt.figure(figsize=(12,4))
sns.countplot(
    data=df, 
    x='pickup_borough', 
    order=df['pickup_borough'].value_counts().index,
    palette='viridis'  # Clean, modern color spectrum
)
plt.title('Number of Trips by Pickup Borough', fontsize=14, fontweight='bold')
plt.xlabel('Pickup Borough', fontsize=12)
plt.ylabel('Number of Trips (Count)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# A scatter plot to show the relationship between distance and fare.

plt.figure(figsize=(12,4))
sns.scatterplot(data=df, x='distance', y='fare', hue='pickup_borough', palette='bright', alpha=0.6)
plt.title('Relationship Between Trip Distance and Fare', fontsize=14, fontweight='bold')
plt.xlabel('Distance (miles)', fontsize=12)
plt.ylabel('Fare ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Pickup Borough', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
# A heatmap to visualize the correlation between numerical variables such as distance, fare, tip, tolls, and total.

numerical_cols = ['distance', 'fare', 'tip', 'tolls', 'total']
corr_matrix = df[numerical_cols].corr()
plt.figure(figsize=(12, 4))
sns.heatmap(
    corr_matrix, 
    annot=True,          # Displays the exact correlation number inside each square
    cmap='bwr',     # Preset diverging color scheme (Red = Positive, Blue = Negative)
    fmt=".2f",           # Rounds the displayed numbers to 2 decimal places
    vmin=-1, vmax=1,     # Anchors the color scale strictly from -1 to 1
    linewidths=0.5,      # Adds a thin, clean border between squares
    square=True          # Forces the cells to be perfect squares
)
plt.title('Correlation Matrix of Trip Metrics', fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```

```python
# A pair plot to visualize the pairwise relationships between distance, fare, tip, and total. 

top_zones = df['pickup_zone'].value_counts().head(5).index
filtered_df = df[df['pickup_zone'].isin(top_zones)]
plot_cols = ['distance', 'fare', 'tip', 'total', 'pickup_zone']
g = sns.pairplot(
    data=filtered_df[plot_cols], 
    hue='pickup_zone', 
    palette='bright', 
    corner=True,
    plot_kws={'alpha': 0.5, 's': 15}  # Makes points semi-transparent and smaller
)
g.fig.suptitle('Pairwise Relationships Across Top Pickup Zones', fontsize=14, fontweight='bold', y=1.02)
plt.show()
```

```python
# A violin plot to show the distribution of fare for each payment method.

plt.figure(figsize=(12,4))
sns.violinplot(data=df, x='payment', y='fare', palette='muted', inner='box')
plt.title('Distribution of Fares by Payment Method', fontsize=14, fontweight='bold')
plt.xlabel('Payment Method', fontsize=12)
plt.ylabel('Fare ($)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.ylim(0,50)
plt.tight_layout()
plt.show()
```
